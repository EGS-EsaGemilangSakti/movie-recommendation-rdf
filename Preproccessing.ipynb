#### Extracting Values:

```python
def convert(text):
    l=[]
    for i in ast.literal_eval(text):
        l.append(i["name"])
    return l
```

#### Applying this function

```python
movies.sample()
```

```python
movies["genres"] = movies["genres"].apply(convert)
```

```python
movies["production_countries"] = movies["production_countries"].apply(convert)
```

```python
movies["production_companies"] = movies["production_companies"].apply(convert)
```

#### Extracting Cast:

```python
def convert_cast(text):
    l=[]
    counter=0
    for i in ast.literal_eval(text):
        if counter<3:
            l.append(i["name"])
        else:
            break
        counter+=1
    return l
```

Applying Function to cast column.
Also making cast dataframe for future use

```python
movies["cast"] = movies["cast"].apply(convert_cast)
```

#### Function to fetch director from crew data

```python
def fetch_director(text):
    l=[]
    for i in ast.literal_eval(text):
        if i["job"] == "Director":
            l.append(i["name"])
            break
        
    return l
```

Applying it to crew column. Making Director for future use

```python
movies["crew"] = movies["crew"].apply(fetch_director)
```

```python

movies['Actor_1'] = movies["cast"].apply(lambda x: x[0] if len(x) > 0 else None)
movies['Actor_2'] = movies["cast"].apply(lambda x: x[1] if len(x) > 1 else None)
movies['Actor_3'] = movies["cast"].apply(lambda x: x[2] if len(x) > 2 else None)

movies = movies.drop("cast", axis=1)  
```

```python
movies['genre_1'] = movies["genres"].apply(lambda x: x[0] if len(x) > 0 else None)
movies['genre_2'] = movies["genres"].apply(lambda x: x[1] if len(x) > 1 else None)
movies['genre_3'] = movies["genres"].apply(lambda x: x[2] if len(x) > 2 else None)
movies['genre_4'] = movies["genres"].apply(lambda x: x[3] if len(x) > 3 else None)
movies['genre_5'] = movies["genres"].apply(lambda x: x[4] if len(x) > 4 else None)

movies = movies.drop("genres", axis=1)  
```

```python
movies['Director'] = movies["crew"].apply(lambda x: x[0] if len(x) > 0 else None)
movies = movies.drop("crew", axis=1)  
```

```python
movies['Country_1'] = movies["production_countries"].apply(lambda x: x[0] if len(x) > 0 else None)
movies['Country_2'] = movies["production_countries"].apply(lambda x: x[1] if len(x) > 1 else None)
movies['Country_3'] = movies["production_countries"].apply(lambda x: x[2] if len(x) > 2 else None)
movies['Country_4'] = movies["production_countries"].apply(lambda x: x[3] if len(x) > 3 else None)
movies['Country_5'] = movies["production_countries"].apply(lambda x: x[4] if len(x) > 4 else None)

movies = movies.drop("production_countries", axis=1)  
```
